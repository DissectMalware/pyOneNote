# Parse OneNote Notebook with Section Groups

This notebook demonstrates how to use pyOneNote to parse a OneNote notebook directory and extract sections, **section groups**, pages, and content.

**Note:** OneNote notebooks can be organized with Section Groups (folders containing sections). This example shows how to access both.

# Parse OneNote Notebook Example

This notebook demonstrates how to use pyOneNote to parse a OneNote notebook directory and extract sections, pages, and content.

## Step 1: Import Required Libraries

```python
import sys
from pathlib import Path

# Add pyOneNote to path if needed
pyonenote_path = r'C:\programing\checkout\pyOneNote'
if pyonenote_path not in sys.path:
    sys.path.insert(0, pyonenote_path)

from pyOneNote import OneNoteParser, OneNoteNotebook
import json
```

## Step 2: Define Notebook Path and Find .one Files

```python
# Define the notebook directory
notebook_dir = Path(r'C:\Users\oz303\Oz\MyNotebook')

# Check if directory exists
if not notebook_dir.exists():
    print(f"Error: Directory not found: {notebook_dir}")
else:
    print(f"Notebook directory: {notebook_dir}")
    
    # Find all .one files
    one_files = list(notebook_dir.rglob('*.one'))
    print(f"\nFound {len(one_files)} .one file(s):")
    for f in one_files:
        print(f"  - {f.name}")
```

## Step 3: Parse the First OneNote File

```python
# Parse the first .one file
if one_files:
    one_file = one_files[0]
    print(f"Parsing: {one_file}\n")
    
    parser = OneNoteParser(str(one_file))
    print("✓ File parsed successfully")
else:
    print("No .one files found!")
```

## Step 4: Extract Sections

```python
# Get sections from the notebook
sections = parser.get_sections()

print(f"Found {len(sections)} section(s):\n")
for i, section in enumerate(sections, 1):
    print(f"Section {i}:")
    print(f"  ID: {section.section_id}")
    print(f"  Title: {section.title}")
    print(f"  Pages: {len(section.pages)}")
    print()
```

```python
# Get section groups
section_groups = parser.document.find_section_group_nodes()

print(f"Found {len(section_groups)} section group(s)\\n")

# Show section group details
print("Section Groups:") 
for i, group_node in enumerate(section_groups, 1):
    if hasattr(group_node, 'data') and hasattr(group_node.data, 'body'):
        group_id = str(group_node.data.body.oid)
        print(f"  Group {i}: ID={group_id}")
```

## Step 5b: Extract Section Groups

Section Groups are containers that can hold multiple sections (similar to folders in Windows Explorer).

## Step 5: Extract Pages

```python
# Get all pages
pages = parser.get_pages()

print(f"Found {len(pages)} page(s)\n")

# Show first 5 pages
print("First 5 pages:")
for i, page in enumerate(pages[:5], 1):
    print(f"\nPage {i}:")
    print(f"  ID: {page.page_id}")
    print(f"  Title: {page.title}")
    print(f"  Content length: {len(page.content)} chars")
    if page.content:
        print(f"  Preview: {page.content[:100]}...")
```

## Step 6: Create OneNoteNotebook Object

```python
# Get the complete notebook structure
notebook = parser.get_notebook()

print(f"Notebook Title: {notebook.title}")
print(f"Notebook ID: {notebook.notebook_id}")
print(f"\nStructure:")
print(f"  Sections: {len(notebook.sections)}")
print(f"  Section Groups: {len(notebook.section_groups)}")

# Count total pages
total_pages = sum(len(s.pages) for s in notebook.sections)
print(f"  Total Pages: {total_pages}")
```

## Step 7: Export Notebook to JSON

```python
# Convert notebook to dictionary
notebook_dict = notebook.to_dict()

# Show structure (truncated for display)
print("Notebook structure (keys):")
for key, value in notebook_dict.items():
    if isinstance(value, list):
        print(f"  {key}: {len(value)} items")
    else:
        print(f"  {key}: {value}")

# Optionally save to JSON
# with open('notebook_export.json', 'w', encoding='utf-8') as f:
#     json.dump(notebook_dict, f, indent=2, default=str)
#     print("\n✓ Saved to notebook_export.json")
```

## Step 8: Access Page Content

```python
# Access pages from the first section
if notebook.sections and notebook.sections[0].pages:
    first_section = notebook.sections[0]
    first_page = first_section.pages[0]
    
    print(f"First page in first section:")
    print(f"  Title: {first_page.title}")
    print(f"  Content:")
    print(f"  {first_page.content[:200]}")
    print(f"\n  Full content length: {len(first_page.content)} chars")
else:
    print("No pages found in sections")
```
