# Simple OneNote Notebook Usage

Direct parsing from path with automatic tree population

```python
import sys
sys.path.insert(0, r'C:\programing\checkout\pyOneNote')

from pyOneNote.models import OneNoteNotebook
```

## Method 1: Parse from File Path

Simply pass the path to a .one file

```python
# Create and parse notebook from file path
notebook = OneNoteNotebook(path=r'C:\Users\oz303\Oz\MyNotebook\New Section 1.one')

print(f'Notebook loaded: {notebook.title}')
print(f'Sections: {len(notebook.sections)}')
print(f'Section Groups: {len(notebook.section_groups)}')
print(f'Pages: {sum(len(s.pages) for s in notebook.sections)}')
```

## Method 2: Parse from Directory Path

Pass a directory and it auto-finds the first .one file

```python
# Create and parse from directory (auto-discovers .one files)
notebook = OneNoteNotebook(path=r'C:\Users\oz303\Oz\MyNotebook')

print(f'Notebook loaded: {notebook.title}')
print(f'Sections: {len(notebook.sections)}')
print(f'Section Groups: {len(notebook.section_groups)}')
print(f'Pages: {sum(len(s.pages) for s in notebook.sections)}')
```

## Access the Tree Structure

Navigate through sections, section groups, and pages

```python
# Access sections
print('Sections:')
for section in notebook.sections:
    print(f'  - {section.title}: {len(section.pages)} pages')
```

```python
# Access section groups
print(f'\nSection Groups: {len(notebook.section_groups)}')
for i, group in enumerate(notebook.section_groups, 1):
    print(f'  Group {i}: {group.title or "(no title)"} - {len(group.sections)} sections')
```

```python
# Access pages from the first section
if notebook.sections and notebook.sections[0].pages:
    first_section = notebook.sections[0]
    first_page = first_section.pages[0]
    
    print(f'First page in first section:')
    print(f'  Title: {first_page.title}')
    print(f'  Content preview: {first_page.content[:150]}...')
```

## Export to JSON

Serialize the entire notebook structure

```python
import json

# Convert to dictionary
notebook_dict = notebook.to_dict()

# Show structure summary
print('Notebook structure:')
print(f'  Sections: {len(notebook_dict["sections"])}')
print(f'  Section Groups: {len(notebook_dict["section_groups"])}')
print(f'  Total pages: {sum(len(s["pages"]) for s in notebook_dict["sections"])}')

# Save to file (optional)
# with open('notebook.json', 'w', encoding='utf-8') as f:
#     json.dump(notebook_dict, f, indent=2, default=str)
#     print('\n✓ Saved to notebook.json')
```
